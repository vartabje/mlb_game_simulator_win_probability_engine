# Testing with direct scraping

```python
import pandas as pd
import requests
from io import StringIO

url = "https://baseballsavant.mlb.com/leaderboard/custom"

params = {
    "year": 2023,
    "type": "batter",
    "min": 1,
    "selections": ",".join([
        "pa",
        "ab",
        "xwoba",
        "xslg",
        "exit_velocity_avg",
        "barrel_batted_rate",
        "hard_hit_percent",
        "sweet_spot_percent",
        "oz_swing_percent",
        "whiff_percent",
        "k_percent",
        "bb_percent",

    ]),
    "chart": "false",
    "csv": "true"
}

response = requests.get(url, params=params, timeout=60)
response.raise_for_status()

batter_df = pd.read_csv(StringIO(response.text))

display(batter_df.head())
```
